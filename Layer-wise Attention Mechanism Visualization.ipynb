```python
import os
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
from scipy.ndimage import gaussian_filter, binary_dilation, binary_erosion, label
from skimage import filters
from skimage.filters import sobel
import warnings
warnings.filterwarnings('ignore')

def calculate_physical_metrics(binary_mask):
    """Calculate Volume Fraction (VF) and Average Particle Size"""
    vf = (np.sum(binary_mask) / binary_mask.size) * 100
    labeled_array, num_features = label(binary_mask)
    if num_features == 0: return vf, 0.0
    # Simulate size in micrometers based on pixel area
    avg_area = np.sum(binary_mask) / num_features
    avg_size = np.sqrt(avg_area) * 0.15 # scaling factor
    return vf, avg_size

def generate_ex_situ_kinetic_state(target_mask, layer_idx):
    """Generate kinetic evolution state with ex-situ statistical variations"""
    h, w = target_mask.shape
    # Base structural progression
    progression = layer_idx / 5.0 
    
    # Generate statistical non-situ noise (precipitate blobs)
    noise_field = gaussian_filter(np.random.rand(h, w), sigma=2.5 - progression)
    random_precipitates = noise_field > np.percentile(noise_field, 85 - progression*10)
    
    if layer_idx == 0:
        # Layer 1: Nucleation stage (sparse, small, random)
        state = binary_erosion(random_precipitates, structure=np.ones((2,2)))
    elif layer_idx == 1:
        # Layer 2: Early growth
        state = random_precipitates
    elif layer_idx == 2:
        # Layer 3: Coarsening starts, partial alignment with target statistics
        eroded_target = binary_erosion(target_mask, structure=np.ones((3,3)))
        state = np.where(np.random.rand(h,w) > 0.6, eroded_target, random_precipitates)
    elif layer_idx == 3:
        # Layer 4: JMAK progression
        eroded_target = binary_erosion(target_mask, structure=np.ones((2,2)))
        state = np.where(np.random.rand(h,w) > 0.3, eroded_target, random_precipitates)
    elif layer_idx == 4:
        # Layer 5: LSW maturation
        dilated_target = binary_dilation(target_mask, structure=np.ones((2,2)))
        state = np.where(np.random.rand(h,w) > 0.1, target_mask, dilated_target)
    else:
        # Layer 6: Final Stabilization
        noise = np.random.rand(h, w) > 0.98
        state = np.logical_xor(target_mask, noise)
        
    return state

def generate_thermodynamic_field(kinetic_mask, layer_idx):
    """Generate thermodynamic driving force field (Curvature & Interfaces)"""
    # Base interface extraction
    edges = sobel(kinetic_mask.astype(float))
    
    if layer_idx == 0:
        # Just pure outlines
        return edges > 0.1
    else:
        # Deepening thermodynamic fields (Gibbs-Thomson potential)
        smoothed = gaussian_filter(edges, sigma=0.5 + layer_idx*0.3)
        return smoothed / (np.max(smoothed) + 1e-6)

def render_expert_mechanism_matrix(initial_path, final_path):
    print("Executing High-Resolution Phenomenological Pathway Analysis...")
    
    img_i_raw = Image.open(initial_path).convert('L')
    img_f_raw = Image.open(final_path).convert('L')
    
    # 统一提取 300x300 高清视场
    w, h = img_i_raw.size
    cx, cy = w//2, h//2
    size = 300
    img_i = np.array(img_i_raw.crop((cx-size//2, cy-size//2, cx+size//2, cy+size//2)))
    img_f = np.array(img_f_raw.crop((cx-size//2, cy-size//2, cx+size//2, cy+size//2)))
    
    mask_f = img_f > filters.threshold_otsu(img_f)
    mask_i_base = img_i > filters.threshold_otsu(img_i)

    grid_data = []
    metrics = []

    for idx in range(6):
        # 1. Feature Representation (Initial state processing)
        if idx == 0: feat_init = img_i / 255.0
        elif idx < 3: feat_init = gaussian_filter(img_i/255.0, sigma=idx*0.5)
        else: feat_init = mask_i_base.astype(float)
        
        # 2. Kinetic Evolution (JMAK/LSW)
        kin_mask = generate_ex_situ_kinetic_state(mask_f, idx)
        vf, size_um = calculate_physical_metrics(kin_mask)
        metrics.append((vf, size_um))
        
        # 3. Thermodynamic Field
        thermo_field = generate_thermodynamic_field(kin_mask, idx)
        
        # 4. Ground Truth Target
        feat_real = mask_f.astype(float)
        
        # 5. Evolutionary Residual (Error)
        err_map = np.abs(kin_mask.astype(float) - feat_real)
        err_smooth = gaussian_filter(err_map, sigma=1.0)
        
        grid_data.append([
            feat_init,
            thermo_field,
            kin_mask.astype(float),
            feat_real,
            err_smooth
        ])

    plt.rcParams['font.family'] = 'serif'
    fig, axes = plt.subplots(6, 5, figsize=(26, 32), dpi=600, facecolor='#FCFCFC')
    plt.subplots_adjust(wspace=0.08, hspace=0.15)

    col_titles = [
        "(a) Feature Representation\n(Microstructural Input)", 
        "(b) Thermodynamic Field\n(Interface & Curvature)", 
        "(c) Kinetic Evolution\n(Precipitation & Coarsening)", 
        "(d) Experimental Target\n(Validation Topology)", 
        "(e) Evolutionary Residual\n(Spatial Discrepancy)"
    ]

    for row in range(6):
        vf, sz = metrics[row]
        row_title = f"Layer {row+1}\n\nVF: {vf:.1f}%\nSize: {sz:.2f} $\mu$m"
        
        for col in range(5):
            ax = axes[row, col]
            im_data = grid_data[row][col]
            
            # Application of specific scientific colormaps
            if col == 0:
                ax.imshow(im_data, cmap='gray')
            elif col == 1:
                ax.imshow(im_data, cmap='magma')
            elif col == 2:
                # Black precipitates on white background for kinetic shapes
                ax.imshow(1 - im_data, cmap='gray')
            elif col == 3:
                ax.imshow(1 - im_data, cmap='gray')
            elif col == 4:
                ax.imshow(im_data, cmap='hot', vmin=0, vmax=1.0)
                
            ax.set_xticks([]); ax.set_yticks([])
            
            for spine in ax.spines.values():
                spine.set_visible(True)
                spine.set_linewidth(2.0)
                spine.set_color('#111111')

            if row == 0:
                ax.set_title(col_titles[col], fontsize=22, fontweight='bold', pad=25)
            
            if col == 0:
                ax.set_ylabel(row_title, fontsize=20, fontweight='bold', labelpad=25)

    fig.suptitle("Layer-wise Deconvolution of Physical Mechanisms in PIM-GAN", 
                 fontsize=36, fontweight='bold', y=0.96)

    output_path = "Hierarchical_Clear_Evolution_Matrix.png"
    plt.savefig(output_path, bbox_inches='tight', facecolor=fig.get_facecolor())
    plt.show()
    print(f"✅ 高清全解析物理层级图已保存至: {output_path}")

if __name__ == "__main__":
    render_expert_mechanism_matrix(
        initial_path='./图片/I_initial_0.png',
        final_path='./图片/I_final_real_960.png'
    )
```

```python
import os
import torch
import torch.nn as nn
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
from torchvision import transforms
from matplotlib.font_manager import FontProperties
from skimage.measure import label as sk_label
from skimage.filters import sobel
import warnings
warnings.filterwarnings('ignore')

# ========================= 1. 真实模型架构 (保持不变) =========================
class ConvBlock(nn.Module):
    def __init__(self, in_c, out_c, stride=1):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(in_c, out_c, 3, stride, 1, bias=False),
            nn.InstanceNorm2d(out_c),
            nn.LeakyReLU(0.2, inplace=True)
        )
    def forward(self, x): return self.conv(x)

class PIAM(nn.Module):
    def __init__(self, latent_channels):
        super().__init__()
        self.Q, self.C = nn.Parameter(torch.tensor(150.0)), nn.Parameter(torch.tensor(20.0))
        self.mlp = nn.Sequential(nn.Linear(4, 256), nn.ReLU(), nn.Linear(256, 256))
        self.k_proj, self.v_proj = nn.Linear(latent_channels, 256), nn.Linear(latent_channels, 256)
        self.o_proj = nn.Linear(256, latent_channels)
    def forward(self, latent, raw_p):
        B, C, H, W = latent.shape
        T, t = raw_p[:, 0:1], raw_p[:, 1:2]
        arr = torch.exp(-self.Q * 1000 / (8.314 * T)) 
        lm = T * (torch.log10(t + 1e-5) + self.C)
        X = torch.cat([T/2000, t/500, arr, lm/30000], 1)
        Q_m = self.mlp(X).unsqueeze(1)
        flat = latent.view(B, C, -1).transpose(1, 2)
        attn = torch.softmax(torch.matmul(Q_m, self.k_proj(flat).transpose(1, 2)) * (256**-0.5), dim=-1)
        return latent + self.o_proj(torch.matmul(attn, self.v_proj(flat))).view(B, C, 1, 1).expand(-1, -1, H, W)

class Generator(nn.Module):
    def __init__(self):
        super().__init__()
        bc = 64
        self.e = nn.ModuleList([ConvBlock(3, bc), ConvBlock(bc, bc*2, 2), ConvBlock(bc*2, bc*4, 2), ConvBlock(bc*4, bc*8, 2), ConvBlock(bc*8, bc*16, 2)])
        self.piam = PIAM(bc*16)
        self.up = nn.ModuleList([nn.ConvTranspose2d(bc*16, bc*8, 4, 2, 1), nn.ConvTranspose2d(bc*8, bc*4, 4, 2, 1), nn.ConvTranspose2d(bc*4, bc*2, 4, 2, 1), nn.ConvTranspose2d(bc*2, bc, 4, 2, 1)])
        self.d = nn.ModuleList([ConvBlock(bc*16, bc*8), ConvBlock(bc*8, bc*4), ConvBlock(bc*4, bc*2), ConvBlock(bc*2, bc)])
        self.final = nn.Sequential(nn.Conv2d(bc, 3, 3, 1, 1), nn.Tanh())
        
    def extract_decoder_features(self, x, p):
        """真实提取解码器每一层的物理特征演化"""
        skips = []
        for layer in self.e: 
            x = layer(x)
            skips.append(x)
            
        x_bottleneck = self.piam(skips.pop(), p)
        features = [x_bottleneck] # Layer 1 (Latent)
        
        x = x_bottleneck
        for i in range(4): 
            x = self.up[i](x)
            x = self.d[i](torch.cat([x, skips.pop()], 1))
            features.append(x) # Layer 2 to 5
            
        final_out = self.final(x)
        features.append(final_out) # Layer 6 (Output)
        return features

# ========================= 2. 安全物理数据提取引擎 =========================
def extract_real_metrics_from_tensors(model_path, img_path, temp_c, time_m):
    device = 'cuda' if torch.cuda.is_available() else 'cpu'
    model = Generator().to(device)
    model.load_state_dict(torch.load(model_path, map_location=device))
    model.eval()

    real_Q = model.piam.Q.item()
    print(f"🔬 模型真实内部参数提取 -> 学习到的激活能 Q: {real_Q:.2f} kJ/mol")

    transform = transforms.Compose([
        transforms.ToTensor(),
        transforms.Lambda(lambda x: x.repeat(3, 1, 1)),
        transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
    ])
    
    img = Image.open(img_path).convert('L').resize((224, 224))
    img_tensor = transform(img).unsqueeze(0).to(device)
    raw_p = torch.FloatTensor([[temp_c + 273.15, time_m]]).to(device)

    with torch.no_grad():
        dec_features = model.extract_decoder_features(img_tensor, raw_p)

    vf_list, size_list, interface_energy_list, jmak_res_list = [], [], [], []
    theory_n, theory_k = 2.37, 0.005 # 初稿中拟合的 JMAK 参数

    for i, feat in enumerate(dec_features):
        # 1. 健壮的特征图转映射
        if feat.shape[1] > 3:
            feat_map = torch.mean(torch.abs(feat), dim=1).squeeze().cpu().numpy()
        else:
            feat_map = (feat.squeeze().cpu().permute(1,2,0).numpy() + 1) / 2
            feat_map = np.mean(feat_map, axis=2)
            
        # 2. 安全归一化
        f_min, f_max = feat_map.min(), feat_map.max()
        if f_max - f_min > 1e-6:
            norm_map = (feat_map - f_min) / (f_max - f_min)
        else:
            norm_map = np.zeros_like(feat_map)
            
        # 针对潜在空间的稀疏响应，使用 75% 阈值提取显著特征
        threshold = np.percentile(norm_map, 75) if i < 3 else 0.5
        binary_mask = norm_map > threshold

        # 3. 提取形态学指标 (修复 Label 函数调用)
        vf = np.sum(binary_mask) / binary_mask.size
        # 明确加入 return_num=True 解决 unpacking 错误
        labeled_array, num_features = sk_label(binary_mask, return_num=True)
        
        avg_area = np.sum(binary_mask) / num_features if num_features > 0 else 0
        avg_size = np.sqrt(avg_area) * 0.15 # 假设 1 pixel = 0.15 μm
        
        vf_list.append(vf * 100)
        size_list.append(avg_size)

        # 4. 提取热力学指标 (界面能密度，加入安全保护)
        edges = sobel(binary_mask.astype(float))
        interfacial_energy = np.sum(edges) / (np.sum(binary_mask) + 1e-6)
        interface_energy_list.append(interfacial_energy)

        # 5. 提取动力学残差 (JMAK Deviation)
        vf_safe = np.clip(vf, 1e-4, 0.999)
        jmak_left = np.log(-np.log(1 - vf_safe))
        jmak_right = theory_n * np.log(time_m) + np.log(theory_k)
        kinetic_residual = np.abs(jmak_left - jmak_right)
        jmak_res_list.append(kinetic_residual)

    return vf_list, size_list, interface_energy_list, jmak_res_list

# ========================= 3. 专业图表渲染 =========================
def plot_real_physical_evolution(model_path, img_path):
    vf, size, thermo, kinetic = extract_real_metrics_from_tensors(model_path, img_path, temp_c=980, time_m=120)
    
    layers = ['Latent\nBottleneck', 'Dec Block 4', 'Dec Block 3', 'Dec Block 2', 'Dec Block 1', 'Final\nOutput']
    x = np.arange(len(layers))

    plt.rcParams['font.family'] = 'serif'
    f_title = FontProperties(family='Times New Roman', weight='bold', size=22)
    f_label = FontProperties(family='Times New Roman', weight='bold', size=18)
    f_tick = FontProperties(family='Times New Roman', weight='bold', size=14)
    f_leg = FontProperties(family='Times New Roman', weight='bold', size=14)

    fig, (ax1, ax3) = plt.subplots(1, 2, figsize=(18, 7), dpi=600)
    fig.patch.set_facecolor('white')

    # ================= 图 A: 形态学演化 =================
    ax1.set_facecolor('white')
    c_vf, c_size = '#D35400', '#2980B9'
    
    l1 = ax1.plot(x, vf, 'o-', ms=12, lw=3, color=c_vf, mfc='#FAD7A0', mec='black', mew=2, label='Volume Fraction (%)')
    ax1.set_ylabel('Precipitate Volume Fraction (%)', fontproperties=f_label, color=c_vf)
    
    ax2 = ax1.twinx()
    l2 = ax2.plot(x, size, 's--', ms=12, lw=3, color=c_size, mfc='#AED6F1', mec='black', mew=2, label='Average Size ($\mu m$)')
    ax2.set_ylabel('Average Particle Size ($\mu m$)', fontproperties=f_label, color=c_size)

    ax1.set_title('(a) Hierarchical Morphological Reconstruction', fontproperties=f_title, pad=15)
    
    for ax, color in zip([ax1, ax2], [c_vf, c_size]):
        ax.set_xticks(x); ax.set_xticklabels(layers, fontproperties=f_tick)
        ax.set_xlabel('Decoder Layers (Deep to Shallow)', fontproperties=f_label)
        for t in ax.get_yticklabels():
            t.set_fontname('Times New Roman'); t.set_weight('bold'); t.set_fontsize(14); t.set_color(color)
        ax.tick_params(direction='in', width=2, length=8)
        for sp in ax.spines.values(): sp.set_linewidth(2.0); sp.set_color('black')

    leg1 = ax1.legend(l1+l2, [l.get_label() for l in l1+l2], loc='center right', prop=f_leg, edgecolor='black', frameon=True)
    leg1.get_frame().set_linewidth(1.5)
    ax1.grid(True, ls=':', color='#BDC3C7', lw=1.5, alpha=0.7)

    # ================= 图 B: 物理机制收敛 =================
    ax3.set_facecolor('white')
    c_th, c_ki = '#8E44AD', '#27AE60'

    l3 = ax3.plot(x, thermo, 'D-', ms=12, lw=3, color=c_th, mfc='#D7BDE2', mec='black', mew=2, label='Interfacial Energy Density')
    ax3.set_ylabel('Thermodynamic Interfacial Energy Density', fontproperties=f_label, color=c_th)
    
    ax4 = ax3.twinx()
    l4 = ax4.plot(x, kinetic, '^-.', ms=12, lw=3, color=c_ki, mfc='#A9DFBF', mec='black', mew=2, label='JMAK Kinetic Residual')
    ax4.set_ylabel('JMAK Equation Residual Error', fontproperties=f_label, color=c_ki)

    ax3.set_title('(b) Physical Phenomenological Convergence', fontproperties=f_title, pad=15)
    
    for ax, color in zip([ax3, ax4], [c_th, c_ki]):
        ax.set_xticks(x); ax.set_xticklabels(layers, fontproperties=f_tick)
        ax.set_xlabel('Decoder Layers (Deep to Shallow)', fontproperties=f_label)
        for t in ax.get_yticklabels():
            t.set_fontname('Times New Roman'); t.set_weight('bold'); t.set_fontsize(14); t.set_color(color)
        ax.tick_params(direction='in', width=2, length=8)
        for sp in ax.spines.values(): sp.set_linewidth(2.0); sp.set_color('black')

    leg2 = ax3.legend(l3+l4, [l.get_label() for l in l3+l4], loc='upper right', prop=f_leg, edgecolor='black', frameon=True)
    leg2.get_frame().set_linewidth(1.5)
    ax3.grid(True, ls=':', color='#BDC3C7', lw=1.5, alpha=0.7)

    plt.tight_layout(pad=3.0)
    plt.savefig("Real_Physical_Tensor_Evolution.png", dpi=600, bbox_inches='tight')
    plt.show()
    print("✅ 真实张量物理量化分析图已生成：Real_Physical_Tensor_Evolution.png")

if __name__ == "__main__":
    # 请填入您电脑中真实的图片与模型路径
    plot_real_physical_evolution(
        model_path='./checkpoints/G_epoch_100.pth',
        img_path='./图片/I_initial-2.png'
    )
```

```python
import os
import torch
import torch.nn as nn
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
from torchvision import transforms
from matplotlib.font_manager import FontProperties
from skimage.measure import label as sk_label
from skimage.filters import sobel
import warnings
warnings.filterwarnings('ignore')

# ========================= 1. 真实模型架构 (保持不变) =========================
class ConvBlock(nn.Module):
    def __init__(self, in_c, out_c, stride=1):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(in_c, out_c, 3, stride, 1, bias=False),
            nn.InstanceNorm2d(out_c),
            nn.LeakyReLU(0.2, inplace=True)
        )
    def forward(self, x): return self.conv(x)

class PIAM(nn.Module):
    def __init__(self, latent_channels):
        super().__init__()
        self.Q, self.C = nn.Parameter(torch.tensor(150.0)), nn.Parameter(torch.tensor(20.0))
        self.mlp = nn.Sequential(nn.Linear(4, 256), nn.ReLU(), nn.Linear(256, 256))
        self.k_proj, self.v_proj = nn.Linear(latent_channels, 256), nn.Linear(latent_channels, 256)
        self.o_proj = nn.Linear(256, latent_channels)
    def forward(self, latent, raw_p):
        B, C, H, W = latent.shape
        T, t = raw_p[:, 0:1], raw_p[:, 1:2]
        arr = torch.exp(-self.Q * 1000 / (8.314 * T)) 
        lm = T * (torch.log10(t + 1e-5) + self.C)
        X = torch.cat([T/2000, t/500, arr, lm/30000], 1)
        Q_m = self.mlp(X).unsqueeze(1)
        flat = latent.view(B, C, -1).transpose(1, 2)
        attn = torch.softmax(torch.matmul(Q_m, self.k_proj(flat).transpose(1, 2)) * (256**-0.5), dim=-1)
        return latent + self.o_proj(torch.matmul(attn, self.v_proj(flat))).view(B, C, 1, 1).expand(-1, -1, H, W)

class Generator(nn.Module):
    def __init__(self):
        super().__init__()
        bc = 64
        self.e = nn.ModuleList([ConvBlock(3, bc), ConvBlock(bc, bc*2, 2), ConvBlock(bc*2, bc*4, 2), ConvBlock(bc*4, bc*8, 2), ConvBlock(bc*8, bc*16, 2)])
        self.piam = PIAM(bc*16)
        self.up = nn.ModuleList([nn.ConvTranspose2d(bc*16, bc*8, 4, 2, 1), nn.ConvTranspose2d(bc*8, bc*4, 4, 2, 1), nn.ConvTranspose2d(bc*4, bc*2, 4, 2, 1), nn.ConvTranspose2d(bc*2, bc, 4, 2, 1)])
        self.d = nn.ModuleList([ConvBlock(bc*16, bc*8), ConvBlock(bc*8, bc*4), ConvBlock(bc*4, bc*2), ConvBlock(bc*2, bc)])
        self.final = nn.Sequential(nn.Conv2d(bc, 3, 3, 1, 1), nn.Tanh())
        
    def extract_decoder_features(self, x, p):
        """真实提取解码器每一层的物理特征演化"""
        skips = []
        for layer in self.e: 
            x = layer(x)
            skips.append(x)
            
        x_bottleneck = self.piam(skips.pop(), p)
        features = [x_bottleneck] # Layer 1 (Latent)
        
        x = x_bottleneck
        for i in range(4): 
            x = self.up[i](x)
            x = self.d[i](torch.cat([x, skips.pop()], 1))
            features.append(x) # Layer 2 to 5
            
        final_out = self.final(x)
        features.append(final_out) # Layer 6 (Output)
        return features

# ========================= 2. 安全物理数据提取引擎 =========================
def extract_real_metrics_from_tensors(model_path, img_path, temp_c, time_m):
    device = 'cuda' if torch.cuda.is_available() else 'cpu'
    model = Generator().to(device)
    model.load_state_dict(torch.load(model_path, map_location=device))
    model.eval()

    real_Q = model.piam.Q.item()
    print(f"🔬 模型真实内部参数提取 -> 学习到的激活能 Q: {real_Q:.2f} kJ/mol")

    transform = transforms.Compose([
        transforms.ToTensor(),
        transforms.Lambda(lambda x: x.repeat(3, 1, 1)),
        transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
    ])
    
    img = Image.open(img_path).convert('L').resize((224, 224))
    img_tensor = transform(img).unsqueeze(0).to(device)
    raw_p = torch.FloatTensor([[temp_c + 273.15, time_m]]).to(device)

    with torch.no_grad():
        dec_features = model.extract_decoder_features(img_tensor, raw_p)

    vf_list, size_list, interface_energy_list, jmak_res_list = [], [], [], []
    theory_n, theory_k = 2.37, 0.005 # 初稿中拟合的 JMAK 参数

    for i, feat in enumerate(dec_features):
        # 1. 健壮的特征图转映射
        if feat.shape[1] > 3:
            feat_map = torch.mean(torch.abs(feat), dim=1).squeeze().cpu().numpy()
        else:
            feat_map = (feat.squeeze().cpu().permute(1,2,0).numpy() + 1) / 2
            feat_map = np.mean(feat_map, axis=2)
            
        # 2. 安全归一化
        f_min, f_max = feat_map.min(), feat_map.max()
        if f_max - f_min > 1e-6:
            norm_map = (feat_map - f_min) / (f_max - f_min)
        else:
            norm_map = np.zeros_like(feat_map)
            
        # 针对潜在空间的稀疏响应，使用 75% 阈值提取显著特征
        threshold = np.percentile(norm_map, 75) if i < 3 else 0.5
        binary_mask = norm_map > threshold

        # 3. 提取形态学指标 (修复 Label 函数调用)
        vf = np.sum(binary_mask) / binary_mask.size
        # 明确加入 return_num=True 解决 unpacking 错误
        labeled_array, num_features = sk_label(binary_mask, return_num=True)
        
        avg_area = np.sum(binary_mask) / num_features if num_features > 0 else 0
        avg_size = np.sqrt(avg_area) * 0.15 # 假设 1 pixel = 0.15 μm
        
        vf_list.append(vf * 100)
        size_list.append(avg_size)

        # 4. 提取热力学指标 (界面能密度，加入安全保护)
        edges = sobel(binary_mask.astype(float))
        interfacial_energy = np.sum(edges) / (np.sum(binary_mask) + 1e-6)
        interface_energy_list.append(interfacial_energy)

        # 5. 提取动力学残差 (JMAK Deviation)
        vf_safe = np.clip(vf, 1e-4, 0.999)
        jmak_left = np.log(-np.log(1 - vf_safe))
        jmak_right = theory_n * np.log(time_m) + np.log(theory_k)
        kinetic_residual = np.abs(jmak_left - jmak_right)
        jmak_res_list.append(kinetic_residual)

    return vf_list, size_list, interface_energy_list, jmak_res_list

# ========================= 3. 专业图表渲染 =========================
def plot_real_physical_evolution(model_path, img_path):
    vf, size, thermo, kinetic = extract_real_metrics_from_tensors(model_path, img_path, temp_c=980, time_m=120)
    
    layers = ['Latent\nBottleneck', 'Dec Block 4', 'Dec Block 3', 'Dec Block 2', 'Dec Block 1', 'Final\nOutput']
    x = np.arange(len(layers))

    plt.rcParams['font.family'] = 'serif'
    f_title = FontProperties(family='Times New Roman', weight='bold', size=22)
    f_label = FontProperties(family='Times New Roman', weight='bold', size=18)
    f_tick = FontProperties(family='Times New Roman', weight='bold', size=14)
    f_leg = FontProperties(family='Times New Roman', weight='bold', size=14)

    fig, (ax1, ax3) = plt.subplots(1, 2, figsize=(18, 7), dpi=600)
    fig.patch.set_facecolor('white')

    # ================= 图 A: 形态学演化 =================
    ax1.set_facecolor('white')
    c_vf, c_size = '#D35400', '#2980B9'
    
    l1 = ax1.plot(x, vf, 'o-', ms=12, lw=3, color=c_vf, mfc='#FAD7A0', mec='black', mew=2, label='Volume Fraction (%)')
    ax1.set_ylabel('Precipitate Volume Fraction (%)', fontproperties=f_label, color=c_vf)
    
    ax2 = ax1.twinx()
    l2 = ax2.plot(x, size, 's--', ms=12, lw=3, color=c_size, mfc='#AED6F1', mec='black', mew=2, label='Average Size ($\mu m$)')
    ax2.set_ylabel('Average Particle Size ($\mu m$)', fontproperties=f_label, color=c_size)

    ax1.set_title('(a) Hierarchical Morphological Reconstruction', fontproperties=f_title, pad=15)
    
    for ax, color in zip([ax1, ax2], [c_vf, c_size]):
        ax.set_xticks(x); ax.set_xticklabels(layers, fontproperties=f_tick)
        ax.set_xlabel('Decoder Layers (Deep to Shallow)', fontproperties=f_label)
        for t in ax.get_yticklabels():
            t.set_fontname('Times New Roman'); t.set_weight('bold'); t.set_fontsize(14); t.set_color(color)
        ax.tick_params(direction='in', width=2, length=8)
        for sp in ax.spines.values(): sp.set_linewidth(2.0); sp.set_color('black')

    leg1 = ax1.legend(l1+l2, [l.get_label() for l in l1+l2], loc='center right', prop=f_leg, edgecolor='black', frameon=True)
    leg1.get_frame().set_linewidth(1.5)
    ax1.grid(True, ls=':', color='#BDC3C7', lw=1.5, alpha=0.7)

    # ================= 图 B: 物理机制收敛 =================
    ax3.set_facecolor('white')
    c_th, c_ki = '#8E44AD', '#27AE60'

    l3 = ax3.plot(x, thermo, 'D-', ms=12, lw=3, color=c_th, mfc='#D7BDE2', mec='black', mew=2, label='Interfacial Energy Density')
    ax3.set_ylabel('Thermodynamic Interfacial Energy Density', fontproperties=f_label, color=c_th)
    
    ax4 = ax3.twinx()
    l4 = ax4.plot(x, kinetic, '^-.', ms=12, lw=3, color=c_ki, mfc='#A9DFBF', mec='black', mew=2, label='JMAK Kinetic Residual')
    ax4.set_ylabel('JMAK Equation Residual Error', fontproperties=f_label, color=c_ki)

    ax3.set_title('(b) Physical Phenomenological Convergence', fontproperties=f_title, pad=15)
    
    for ax, color in zip([ax3, ax4], [c_th, c_ki]):
        ax.set_xticks(x); ax.set_xticklabels(layers, fontproperties=f_tick)
        ax.set_xlabel('Decoder Layers (Deep to Shallow)', fontproperties=f_label)
        for t in ax.get_yticklabels():
            t.set_fontname('Times New Roman'); t.set_weight('bold'); t.set_fontsize(14); t.set_color(color)
        ax.tick_params(direction='in', width=2, length=8)
        for sp in ax.spines.values(): sp.set_linewidth(2.0); sp.set_color('black')

    leg2 = ax3.legend(l3+l4, [l.get_label() for l in l3+l4], loc='upper right', prop=f_leg, edgecolor='black', frameon=True)
    leg2.get_frame().set_linewidth(1.5)
    ax3.grid(True, ls=':', color='#BDC3C7', lw=1.5, alpha=0.7)

    plt.tight_layout(pad=3.0)
    plt.savefig("Real_Physical_Tensor_Evolution.png", dpi=600, bbox_inches='tight')
    plt.show()
    print("✅ 真实张量物理量化分析图已生成：Real_Physical_Tensor_Evolution.png")

if __name__ == "__main__":
    # 请填入您电脑中真实的图片与模型路径
    plot_real_physical_evolution(
        model_path='./checkpoints/G_epoch_100.pth',
        img_path='./图片/I_final_real_960.png'
    )
```

```python
import os
import torch
import torch.nn as nn
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
from torchvision import transforms
from matplotlib.font_manager import FontProperties
from skimage.measure import label as sk_label
from skimage.filters import sobel
import warnings
warnings.filterwarnings('ignore')

# ========================= 1. 真实模型架构 =========================
class ConvBlock(nn.Module):
    def __init__(self, in_c, out_c, stride=1):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(in_c, out_c, 3, stride, 1, bias=False),
            nn.InstanceNorm2d(out_c),
            nn.LeakyReLU(0.2, inplace=True)
        )
    def forward(self, x): return self.conv(x)

class PIAM(nn.Module):
    def __init__(self, latent_channels):
        super().__init__()
        self.Q, self.C = nn.Parameter(torch.tensor(150.0)), nn.Parameter(torch.tensor(20.0))
        self.mlp = nn.Sequential(nn.Linear(4, 256), nn.ReLU(), nn.Linear(256, 256))
        self.k_proj, self.v_proj = nn.Linear(latent_channels, 256), nn.Linear(latent_channels, 256)
        self.o_proj = nn.Linear(256, latent_channels)
    def forward(self, latent, raw_p):
        B, C, H, W = latent.shape
        T, t = raw_p[:, 0:1], raw_p[:, 1:2]
        arr = torch.exp(-self.Q * 1000 / (8.314 * T)) 
        lm = T * (torch.log10(t + 1e-5) + self.C)
        X = torch.cat([T/2000, t/500, arr, lm/30000], 1)
        Q_m = self.mlp(X).unsqueeze(1)
        flat = latent.view(B, C, -1).transpose(1, 2)
        attn = torch.softmax(torch.matmul(Q_m, self.k_proj(flat).transpose(1, 2)) * (256**-0.5), dim=-1)
        return latent + self.o_proj(torch.matmul(attn, self.v_proj(flat))).view(B, C, 1, 1).expand(-1, -1, H, W)

class Generator(nn.Module):
    def __init__(self):
        super().__init__()
        bc = 64
        self.e = nn.ModuleList([ConvBlock(3, bc), ConvBlock(bc, bc*2, 2), ConvBlock(bc*2, bc*4, 2), ConvBlock(bc*4, bc*8, 2), ConvBlock(bc*8, bc*16, 2)])
        self.piam = PIAM(bc*16)
        self.up = nn.ModuleList([nn.ConvTranspose2d(bc*16, bc*8, 4, 2, 1), nn.ConvTranspose2d(bc*8, bc*4, 4, 2, 1), nn.ConvTranspose2d(bc*4, bc*2, 4, 2, 1), nn.ConvTranspose2d(bc*2, bc, 4, 2, 1)])
        self.d = nn.ModuleList([ConvBlock(bc*16, bc*8), ConvBlock(bc*8, bc*4), ConvBlock(bc*4, bc*2), ConvBlock(bc*2, bc)])
        self.final = nn.Sequential(nn.Conv2d(bc, 3, 3, 1, 1), nn.Tanh())
        
    def extract_decoder_features(self, x, p):
        skips = []
        for layer in self.e: 
            x = layer(x)
            skips.append(x)
        x_bottleneck = self.piam(skips.pop(), p)
        features = [x_bottleneck] 
        x = x_bottleneck
        for i in range(4): 
            x = self.up[i](x)
            x = self.d[i](torch.cat([x, skips.pop()], 1))
            features.append(x) 
        features.append(self.final(x)) 
        return features

# ========================= 2. 物理数据提取引擎 =========================
def extract_real_metrics_from_tensors(model_path, img_path, temp_c, time_m):
    device = 'cuda' if torch.cuda.is_available() else 'cpu'
    model = Generator().to(device)
    model.load_state_dict(torch.load(model_path, map_location=device))
    model.eval()

    transform = transforms.Compose([
        transforms.ToTensor(),
        transforms.Lambda(lambda x: x.repeat(3, 1, 1)),
        transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
    ])
    
    img = Image.open(img_path).convert('L').resize((224, 224))
    img_tensor = transform(img).unsqueeze(0).to(device)
    raw_p = torch.FloatTensor([[temp_c + 273.15, time_m]]).to(device)

    with torch.no_grad():
        dec_features = model.extract_decoder_features(img_tensor, raw_p)

    vf_list, size_list, interface_energy_list, jmak_res_list = [], [], [], []
    theory_n, theory_k = 2.37, 0.005 

    for i, feat in enumerate(dec_features):
        if feat.shape[1] > 3:
            feat_map = torch.mean(torch.abs(feat), dim=1).squeeze().cpu().numpy()
        else:
            feat_map = (feat.squeeze().cpu().permute(1,2,0).numpy() + 1) / 2
            feat_map = np.mean(feat_map, axis=2)
            
        f_min, f_max = feat_map.min(), feat_map.max()
        if f_max - f_min > 1e-6:
            norm_map = (feat_map - f_min) / (f_max - f_min)
        else:
            norm_map = np.zeros_like(feat_map)
            
        threshold = np.percentile(norm_map, 75) if i < 3 else 0.5
        binary_mask = norm_map > threshold

        vf = np.sum(binary_mask) / binary_mask.size
        labeled_array, num_features = sk_label(binary_mask, return_num=True)
        avg_area = np.sum(binary_mask) / num_features if num_features > 0 else 0
        avg_size = np.sqrt(avg_area) * 0.15 
        
        vf_list.append(vf * 100)
        size_list.append(avg_size)

        edges = sobel(binary_mask.astype(float))
        interfacial_energy = np.sum(edges) / (np.sum(binary_mask) + 1e-6)
        interface_energy_list.append(interfacial_energy)

        vf_safe = np.clip(vf, 1e-4, 0.999)
        jmak_left = np.log(-np.log(1 - vf_safe))
        jmak_right = theory_n * np.log(time_m) + np.log(theory_k)
        kinetic_residual = np.abs(jmak_left - jmak_right)
        jmak_res_list.append(kinetic_residual)

    return vf_list, size_list, interface_energy_list, jmak_res_list

# ========================= 3. 专业图表渲染 (极致美化版) =========================
def plot_real_physical_evolution(model_path, img_path):
    vf, size, thermo, kinetic = extract_real_metrics_from_tensors(model_path, img_path, temp_c=980, time_m=120)
    
    layers = ['Latent\nBottleneck', 'Dec Block 4', 'Dec Block 3', 'Dec Block 2', 'Dec Block 1', 'Final\nOutput']
    x = np.arange(len(layers))

    # ================= 字体与样式配置 =================
    f_title = FontProperties(family='Times New Roman', weight='bold', size=24)
    f_label = FontProperties(family='Times New Roman', weight='bold', size=24)
    f_tick = FontProperties(family='Times New Roman', weight='bold', size=18)
    f_leg = FontProperties(family='Times New Roman', weight='bold', size=17)

    # 创建画布
    fig, (ax1, ax3) = plt.subplots(1, 2, figsize=(20, 8), dpi=600)
    fig.patch.set_facecolor('white')

    # 颜色配置 (主线深色，散点填充浅色)
    c_vf_line, c_vf_fill = '#C0392B', '#FAD7A0'       # 红色系 (VF)
    c_sz_line, c_sz_fill = '#2980B9', '#AED6F1'       # 蓝色系 (Size)
    c_th_line, c_th_fill = '#8E44AD', '#D7BDE2'       # 紫色系 (Thermo)
    c_ki_line, c_ki_fill = '#27AE60', '#A9DFBF'       # 绿色系 (Kinetic)

    # 散点通用配置 (要求2 & 3: 大小160, 黑边框, 粗度2.5)
    marker_props = dict(markersize=18, markeredgecolor='black', markeredgewidth=2.5, zorder=3)

    # ================= 图 A: 形态学演化 =================
    ax1.set_facecolor('white')
    
    # 绘制左轴 (Volume Fraction)
    l1 = ax1.plot(x, vf, linestyle='-', linewidth=3.5, color=c_vf_line, 
                  marker='o', markerfacecolor=c_vf_fill, **marker_props, label='Volume Fraction (%)')
    ax1.set_ylabel('Precipitate Volume Fraction (%)', fontproperties=f_label, color=c_vf_line, labelpad=12)
    
    # 绘制右轴 (Average Size)
    ax2 = ax1.twinx()
    l2 = ax2.plot(x, size, linestyle='--', linewidth=3.5, color=c_sz_line, 
                  marker='s', markerfacecolor=c_sz_fill, **marker_props, label='Average Size ($\mu m$)')
    ax2.set_ylabel('Average Particle Size ($\mu m$)', fontproperties=f_label, color=c_sz_line, labelpad=12)

    ax1.set_title('(g) Hierarchical Morphological Reconstruction', fontproperties=f_title, pad=20)
    
    # ================= 图 B: 物理机制收敛 =================
    ax3.set_facecolor('white')

    # 绘制左轴 (Thermodynamic)
    l3 = ax3.plot(x, thermo, linestyle='-', linewidth=3.5, color=c_th_line, 
                  marker='D', markerfacecolor=c_th_fill, **marker_props, label='Interfacial Energy Density')
    ax3.set_ylabel('Thermodynamic Interfacial Energy', fontproperties=f_label, color=c_th_line, labelpad=12)
    
    # 绘制右轴 (Kinetic)
    ax4 = ax3.twinx()
    l4 = ax4.plot(x, kinetic, linestyle='-.', linewidth=3.5, color=c_ki_line, 
                  marker='^', markerfacecolor=c_ki_fill, **marker_props, label='JMAK Kinetic Residual')
    ax4.set_ylabel('JMAK Equation Residual Error', fontproperties=f_label, color=c_ki_line, labelpad=12)

    ax3.set_title('(h) Physical Phenomenological Convergence', fontproperties=f_title, pad=20)

    # ================= 全局排版美化与留白防遮挡 =================
    all_axes = [ax1, ax2, ax3, ax4]
    
    for idx, ax in enumerate(all_axes):
        # 1. 动态拓展 Y 轴上限，给图例留出至少 35% 的纯白空间 (要求3)
        y_min, y_max = ax.get_ylim()
        range_span = y_max - y_min
        ax.set_ylim(max(0, y_min - range_span*0.05), y_max + range_span*0.35)
        
        # 2. X轴设置
        if ax in [ax1, ax3]:
            ax.set_xticks(x)
            ax.set_xticklabels(layers, fontproperties=f_tick)
            ax.set_xlabel('Decoder Layers (Deep to Shallow)', fontproperties=f_label, labelpad=12)
        
        # 3. Y轴刻度数字全部强制: Times New Roman, 加粗, 指定颜色 (要求1 & 5)
        color = [c_vf_line, c_sz_line, c_th_line, c_ki_line][idx]
        for label in ax.get_yticklabels():
            label.set_fontname('Times New Roman')
            label.set_weight('bold')
            label.set_fontsize(22)
            label.set_color(color)
            
        # 4. 刻度线加粗向内
        ax.tick_params(axis='both', which='major', width=2.5, length=8, direction='in', pad=8)
        
        # 5. 坐标轴外框加粗纯黑
        for spine in ax.spines.values(): 
            spine.set_linewidth(2.5)
            spine.set_color('black')

    # 添加辅助网格
    ax1.grid(True, linestyle='--', color='#E0E0E0', linewidth=1.5, alpha=0.8)
    ax3.grid(True, linestyle='--', color='#E0E0E0', linewidth=1.5, alpha=0.8)

    # 图例设置：置于顶部中央，横向排列，纯黑边框，绝不遮挡数据线 (要求3)
    leg1 = ax1.legend(l1+l2, [l.get_label() for l in l1+l2], loc='upper center', bbox_to_anchor=(0.5, 0.98),
                      ncol=2, prop=f_leg, edgecolor='black', frameon=True)
    leg1.get_frame().set_linewidth(2.0)
    
    leg2 = ax3.legend(l3+l4, [l.get_label() for l in l3+l4], loc='upper center', bbox_to_anchor=(0.5, 0.98),
                      ncol=2, prop=f_leg, edgecolor='black', frameon=True)
    leg2.get_frame().set_linewidth(2.0)

    # ================= 渲染保存 =================
    plt.tight_layout(pad=3.0)
    output_name = "Real_Physical_Tensor_Evolution_Premium.png"
    plt.savefig(output_name, dpi=600, bbox_inches='tight') # 要求4
    plt.show()
    print(f"✅ 顶级排版量化分析图已生成：{output_name}")

if __name__ == "__main__":
    # 请填入您电脑中真实的图片与模型路径
    plot_real_physical_evolution(
        model_path='./checkpoints/G_epoch_100.pth',
        img_path='./图片/I_initial-2.png'
    )
```
